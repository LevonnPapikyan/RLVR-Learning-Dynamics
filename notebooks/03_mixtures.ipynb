# 03. Mixed lengths: relay or stall?

The paper's main claim: training on a **mixture** of lengths creates an implicit curriculum. Short lengths learn first and their progress (a sharper $Q$) makes the next length learnable.

- **Ratio 3**, lengths $\{5, 15, 45\}$: neighbours are close, so the paper predicts a relay.
- **Ratio 7**, lengths $\{5, 35\}$: a big jump, so the paper predicts the long length stays near zero.

Each run takes about 20 to 30 minutes on a laptop CPU (3000 steps). In a mixture each length gets only part of the batch, so take-off comes later than in notebook 02.

```python
import sys; sys.path.append('..')   # the code lives in the repo root
import matplotlib.pyplot as plt
from model import load_box
from train import train, takeoff
from plots import plot_history
box = load_box()
```

```python
Q3, h3 = train(box, [5, 15, 45], steps=3000, eval_every=50)
Q7, h7 = train(box, [5, 35], steps=3000, eval_every=50)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 3.5), sharey=True)
plot_history(h3, 'ratio 3: {5, 15, 45}', ax=axes[0])
plot_history(h7, 'ratio 7: {5, 35}', ax=axes[1])
plt.show()
for name, h in [('ratio 3', h3), ('ratio 7', h7)]:
    print(name, {L: takeoff(h, L) for L in h['success']})
```

**Things to try:** more seeds (take-off times vary), a smoother ladder such as $\{5, 10, 15, 20, 25\}$, or `sharp` (how peaked the box is) at 1.5 and 3.
